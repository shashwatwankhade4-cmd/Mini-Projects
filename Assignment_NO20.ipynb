```python
import pandas as pd

# Load dataset
df = pd.read_csv("ford_car_dataset - ford_car_dataset.csv")

# Independent and dependent variables
X = df.drop("price", axis=1)
Y = df["price"]

print("Shape of X:", X.shape)
print("Shape of Y:", Y.shape)
```

```python
# Find categorical columns
cat_cols = X.select_dtypes(include="object").columns
print(cat_cols)

# One-Hot Encoding
X = pd.get_dummies(X, columns=cat_cols)

# Convert to integer type
X = X.astype(int)

# First 5 rows
print(X.head())
```

```python
from sklearn.preprocessing import StandardScaler

num_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

scaler = StandardScaler()

X[num_cols] = scaler.fit_transform(X[num_cols])

print(X.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.33,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Intercept:")
print(model.intercept_)

print("\nCoefficients:")
print(model.coef_)
```

```python
y_pred = model.predict(X_test)

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(comparison.head(10))
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

The model explains approximately 84% of the variation in car prices.
This indicates that the Linear Regression model performs well on this dataset.
Around 16% of the variation remains unexplained due to factors not captured by the model or because the relationship is not perfectly linear.

```python
import joblib

joblib.dump(model, "LR_ford_car.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(list(X.columns), "columns.pkl")

print("Model and preprocessing objects saved successfully.")
```
